```python
# Simple Password Checker Program
# This program checks if a password meets certain criteria:

print("Welcome to Password Checker!")
password = input("Enter your password: ")

# Now Check length of password
if len(password) < 8:
    print("Too short! Use at least 8 characters")
else:
    print("Good length")

# Now Check for uppercase
if not any(char.isupper() for char in password):
    print("Add at least one uppercase letter")
else:
    print("Contains uppercase")

# Now Check for numbers
if not any(char.isdigit() for char in password):
    print("Add at least one number")
else:
    print("Contains numbers")

# Check for special characters
special_chars = "!@#$%^&*"
if not any(char in special_chars for char in password):
    print(f"Add a special character like {special_chars}")
else:
    print("Contains special characters")
```
