```python
# Simple Grade Tracker
students = []
print("Student Grade Tracker")
print("Commands: 'add', 'view', 'exit'")

while True:
    command = input(" ").lower()
    
    if command == "add":
        name = input("Student name: ")
        grade = input("Grade (A/B/C/D/F): ").upper()
        
        if grade in ("A", "B", "C", "D", "F"):
            students.append((name, grade))
            print(f"Added {name} with grade {grade}")
        else:
            print("Invalid grade. Use A, B, C, D, or F")
    
    elif command == "view":
        if not students:
            print("No students added yet")
        else:
            print("\nStudent Grades:")
            print("-" * 20)
            for student in students:
                print(f"{student[0]}: {student[1]}")
            print("-" * 20)
            
            # Count grades
            grades = [grade for (name, grade) in students]
            print("\nGrade Summary:")
            print(f"A: {grades.count('A')} students")
            print(f"B: {grades.count('B')} students")
            print(f"C: {grades.count('C')} students")
            print(f"D: {grades.count('D')} students")
            print(f"F: {grades.count('F')} students")
    
    elif command == "exit":
        print("Goodbye!")
        break
    
    else:
        print("Invalid command")
```
